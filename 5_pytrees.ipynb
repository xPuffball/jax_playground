# 5. Pytrees
JAX has built-in support for objects that look like dicts of arrays, lists of lists of dicts, nested structures in general basically - these are called Pytrees!

## 5.1 So what is a pytree?
It's a container-like structure built out of a container-like Python objects. Datasets are common pytrees, since they can be something like lists of lists of dicts.

```python
import jax
import jax.numpy as jnp

example_trees = [
    [1, 'a', object()],
    (1, (2, 3), ()),
    [1, {'k1': 2, 'k2': (3, 4)}, 5],
    {'a': 2, 'b': (2, 3)},
    jnp.array([1, 2, 3]),
]

# Print how many leaves the pytrees have.
for pytree in example_trees:
  # This `jax.tree.leaves()` method extracts the flattened leaves from the pytrees.
  leaves = jax.tree.leaves(pytree)
  print(f"{repr(pytree):<45} has {len(leaves)} leaves: {leaves}")
```

Basically, any tree-like structure out of a container-like Python objects can be treated as a pytree in JAX! Classes are considered container-like if they're in the pytree registry, which by default include lists, tuples, and dicts - any objects not in there are treated as a leaf node.

## 5.2 Common pytree functions

`jax.tree.map` - is the most common one! it's basically a python `map` but works over entire pytrees

```python
list_of_lists = [
    [1, 2, 3],
    [1, 2],
    [2, 4]
]

jax.tree.map(lambda x: x ** 2, list_of_lists)

# we can also map n-ary functions over multiple arguments:
another_list_of_lists = list_of_lists
jax.tree.map(lambda x, y: x + y, list_of_lists, another_list_of_lists)
```
