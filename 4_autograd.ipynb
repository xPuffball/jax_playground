# 4. Automatic Differentiation

## 4.1 Taking Gradients with `jax.grad`
We can differentiate a scalar-valued function with the `jax.grad()` transformation:

```python
import jax
import jax.numpy as jnp
from jax import grad

grad_tanh = grad(jnp.tanh)
print(grad_tanh(2.0))
```

note that `jax.grad()` takes a function and returns a function - i.e. it takes $f(x)$ and returns $\nabla f(x)$

and of course we can stack `jax.grad()` to compute higher-order derivatives!
